+ 어떤 종류가 꽃받침이 가장 긴가요?

```python
max_sepal_length_by_species = df.groupby('Species')['SepalLengthCm'].mean().reset_index()
longest_sepal_species = max_sepal_length_by_species.loc[max_sepal_length_by_species['SepalLengthCm'].idxmax()]
print(f"꽃받침 길이가 가장 긴 종류: {longest_sepal_species['Species']} (평균 꽃받침 길이: {longest_sepal_species['SepalLengthCm']:.2f} cm)")
```

+ 어떤 종류가 꽃잎이 가장 넓은가요?

```python
max_petal_width_by_species = df.groupby('Species')['PetalWidthCm'].mean().reset_index()
widest_petal_species = max_petal_width_by_species.loc[max_petal_width_by_species['PetalWidthCm'].idxmax()]
print(f"꽃잎 너비가 가장 넓은 종류: {widest_petal_species['Species']} (평균 꽃잎 너비: {widest_petal_species['PetalWidthCm']:.2f} cm)")
```

+ 결과를 표로 정리해보세요

```python
summary_data = {
    '특성': ['가장 긴 꽃받침', '가장 넓은 꽃잎'],
    '종류': [longest_sepal_species['Species'], widest_petal_species['Species']],
    '평균 값 (cm)': [f"{longest_sepal_species['SepalLengthCm']:.2f}", f"{widest_petal_species['PetalWidthCm']:.2f}"]
}
summary_df = pd.DataFrame(summary_data)
display(summary_df)
```

+ 어떤 종류가 꽃받침 면적이 가장 큰가요?

```python
df['SepalArea'] = df['SepalLengthCm'] * df['SepalWidthCm']
max_sepal_area_by_species = df.groupby('Species')['SepalArea'].mean().reset_index()
largest_sepal_area_species = max_sepal_area_by_species.loc[max_sepal_area_by_species['SepalArea'].idxmax()]
print(f"꽃받침 면적이 가장 큰 종류: {largest_sepal_area_species['Species']} (평균 꽃받침 면적: {largest_sepal_area_species['SepalArea']:.2f} cm²)")
```

+ 어떤 종류가 꽃잎 면적이 가장 큰가요?


```python
df['PetalArea'] = df['PetalLengthCm'] * df['PetalWidthCm']
max_petal_area_by_species = df.groupby('Species')['PetalArea'].mean().reset_index()
largest_petal_area_species = max_petal_area_by_species.loc[max_petal_area_by_species['PetalArea'].idxmax()]

print(f"꽃잎 면적이 가장 큰 종류: {largest_petal_area_species['Species']} (평균 꽃잎 면적: {largest_petal_area_species['PetalArea']:.2f} cm²)")
```

+ 꽃받침 너비가 3.0 이상이고 꽃잎 너비가 1.5 이상인 데이터를 찾아보세요

```python
filtered_df = df[(df['SepalWidthCm'] >= 3.0) & (df['PetalWidthCm'] >= 1.5)]
display(filtered_df)
print(f"\n조건에 맞는 데이터 개수: {len(filtered_df)}개")
```

+ 그 데이터들은 주로 어떤 종류인가요?

```python
species_counts = filtered_df['Species'].value_counts()
display(species_counts.to_frame(name='개수'))

primary_species = species_counts.idxmax()
print(f"\n필터링된 데이터는 주로 '{primary_species}' 종입니다.")
```

+  top 10에 가장 많이 포함된 종류는?

```python
top_10_petal_area = df.nlargest(10, 'PetalArea')
display(top_10_petal_area)

top_10_species_counts = top_10_petal_area['Species'].value_counts()
print(f"\nTop 10에 가장 많이 포함된 종류: {top_10_species_counts.idxmax()} ({top_10_species_counts.max()}개)")
```

+ 가장 적게 포함된 종류는?

```python
least_species_counts = top_10_petal_area['Species'].value_counts()
least_species = least_species_counts.idxmin()

print(f"Top 10에 가장 적게 포함된 종류: {least_species} ({least_species_counts.min()}개)")
```
